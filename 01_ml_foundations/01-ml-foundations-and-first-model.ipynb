## 2. Feature, Target, and Sample

### Sample

A sample is one individual observation or one row in a dataset.

Example:

| Age | Salary | Experience | Loan Approved |
|---|---:|---:|---|
| 25 | 30000 | 2 | Yes |
| 35 | 50000 | 8 | No |

Each row is one **sample**.

### Feature

A feature is an input variable that the machine learning model uses to make a prediction.

In the example above:

- Age
- Salary
- Experience

are the **features**.

### Target

The target is the value that we want the machine learning model to predict.

In the example above:

**Loan Approved** is the target.

### Simple Rule

Features = Input to the model  
Target = What the model predicts  
Sample = One row or observation in the dataset

## 3. Classification vs Regression

Machine learning problems can be divided into different types. Two important supervised learning tasks are classification and regression.

### Classification

Classification is used when the target is a category or class.

Examples:

- Email → Spam / Not Spam
- Patient → Disease / Healthy
- File → Malware / Benign
- Flower → Setosa / Versicolor / Virginica

### Regression

Regression is used when the target is a continuous numerical value.

Examples:

- House → Price
- Weather → Temperature
- Product → Sales Amount
- Employee → Salary

### Easy Shortcut

Category → Classification  
Continuous Number → Regression

## 4. Supervised and Unsupervised Learning

### Supervised Learning

In supervised learning, the training dataset contains both input features and the correct target values.

The model learns the relationship between the features and the target.

Example:

Student Information + Pass/Fail Result  
↓  
Train Model  
↓  
Predict Pass/Fail for a New Student

Supervised learning mainly includes:

- Classification
- Regression

### Unsupervised Learning

In unsupervised learning, the dataset does not contain a target label.

The model tries to find hidden patterns or groups in the data.

Example:

Customer Information  
↓  
Machine Learning Algorithm  
↓  
Different Customer Groups

Common unsupervised learning tasks include:

- Clustering
- Dimensionality Reduction

## 7. Train-Test Split

A machine learning model should be evaluated on data that it did not see during training.

Therefore, we divide the dataset into two parts:

- Training Data
- Testing Data

### Training Data
The model learns patterns from the training data.

### Testing Data
The testing data is used to evaluate how well the trained model performs on unseen data.

A common split is:

- 80% Training Data
- 20% Testing Data

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()

X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Done")
print(X_train.shape)
print(X_test.shape)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

### Why do we use `random_state=42`?

The train-test split is random.

Using a fixed `random_state` makes the result reproducible.

This means that if we run the code again, we will get the same train-test split.

### Why do we use `stratify=y`?

`stratify=y` keeps the class distribution approximately the same in both training and testing datasets.

This is especially useful for classification problems.

```python
import numpy as np

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load dataset
iris = load_iris()

# Features and target
X = iris.data
y = iris.target

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Check shapes
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

# Check class distribution
print("\nTraining class counts:")
print(np.bincount(y_train))

print("\nTesting class counts:")
print(np.bincount(y_test))
```

```python
## 9. Model Evaluation

After training a machine learning model, we need to evaluate its performance.

For classification problems, some common evaluation metrics are:

- Accuracy
- Confusion Matrix
- Precision
- Recall
- F1-score

In this lesson, we will use these metrics to evaluate our Logistic Regression model.
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

### Accuracy

Accuracy measures how many predictions are correct out of all predictions.

Formula:

Accuracy = Correct Predictions / Total Predictions

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

### Confusion Matrix

A confusion matrix shows how many samples were predicted correctly and incorrectly for each class.

- Rows represent actual classes.
- Columns represent predicted classes.
- Diagonal values represent correct predictions.
- Off-diagonal values represent incorrect predictions.

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
)

disp.plot()
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

report = classification_report(
    y_test,
    y_pred,
    target_names=iris.target_names
)

print(report)
```

### Classification Report

The classification report provides more detailed evaluation metrics.

- **Precision**: How many predicted positive cases were actually correct.
- **Recall**: How many actual positive cases were correctly identified.
- **F1-score**: Balance between precision and recall.
- **Support**: Number of actual samples in each class.

```python
import pandas as pd

results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

results["Actual Species"] = results["Actual"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

results["Predicted Species"] = results["Predicted"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

results.head(10)
```

```python
results["Correct"] = results["Actual"] == results["Predicted"]

results.head(10)
```

```python
results[results["Correct"] == False]
```

```python
correct_predictions = (y_test == y_pred).sum()
wrong_predictions = (y_test != y_pred).sum()

print("Correct Predictions:", correct_predictions)
print("Wrong Predictions:", wrong_predictions)
```

### What I Learned

- Accuracy shows the overall percentage of correct predictions.
- A confusion matrix shows class-wise correct and incorrect predictions.
- Precision measures how reliable positive predictions are.
- Recall measures how many actual positive samples were correctly detected.
- F1-score balances precision and recall.
- Classification reports provide detailed class-wise model performance.
- Model evaluation is necessary before trusting predictions.

## Exercise 10: Important Concepts

### Feature

A feature is an input variable used by a machine learning model to make predictions.

Example:

In the Iris dataset:

- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

are features.

### Target

The target is the output that we want the machine learning model to predict.

In the Iris dataset, the flower species is the target.

### Training Data

Training data is the portion of the dataset used by the model to learn patterns.

The model learns from:

`X_train` = Training Features  
`y_train` = Correct Training Labels

### Testing Data

Testing data is unseen data used to evaluate the trained model.

`X_test` contains testing features.

`y_test` contains the actual answers.

### fit()

`fit()` is used to train the machine learning model.

Example:

model.fit(X_train, y_train)

The model learns the relationship between the features and target.

### predict()

`predict()` is used to make predictions using a trained model.

Example:

y_pred = model.predict(X_test)

### Accuracy

Accuracy measures the proportion of predictions that are correct.

Accuracy = Correct Predictions / Total Predictions

For example, if the model correctly predicts 29 out of 30 samples:

Accuracy = 29 / 30 = 96.67%
